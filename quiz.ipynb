```python
# Quiz Game Application using Python Libraries
# Google Colab / Jupyter Notebook

# ---------------- Import Libraries ----------------
import random
import time
import ipywidgets as widgets
from IPython.display import display, clear_output

# ---------------- Question Bank ----------------
quiz_data = [
    {
        "question": "1. What is the capital of India?",
        "options": ["Mumbai", "Delhi", "Chennai", "Kolkata"],
        "answer": "Delhi"
    },
    {
        "question": "2. Which language is used for Data Science?",
        "options": ["Python", "HTML", "CSS", "Java"],
        "answer": "Python"
    },
    {
        "question": "3. Who developed Python?",
        "options": ["Guido van Rossum", "Elon Musk", "Bill Gates", "Mark Zuckerberg"],
        "answer": "Guido van Rossum"
    },
    {
        "question": "4. Which planet is called Red Planet?",
        "options": ["Earth", "Mars", "Venus", "Jupiter"],
        "answer": "Mars"
    },
    {
        "question": "5. Which is fastest animal?",
        "options": ["Lion", "Tiger", "Cheetah", "Horse"],
        "answer": "Cheetah"
    }
]

# Randomize questions
random.shuffle(quiz_data)

# ---------------- Variables ----------------
index = 0
score = 0
start_time = time.time()

# ---------------- UI Widgets ----------------
title = widgets.HTML("""
<h1 style='text-align:center;
font-size:38px;
background:linear-gradient(90deg,#06b6d4,#2563eb,#8b5cf6);
-webkit-background-clip:text;
-webkit-text-fill-color:transparent;'>
🎯 Premium Quiz Challenge
</h1>
""")

question = widgets.HTML()
options = widgets.RadioButtons(options=[])

next_btn = widgets.Button(
    description="Next Question",
    button_style='success'
)

# ---------------- Load Question ----------------
def load_question():
    global index

    if index < len(quiz_data):
        q = quiz_data[index]
        question.value = f"<h3>{q['question']}</h3>"
        options.options = q["options"]
        options.value = None
    else:
        show_result()

# ---------------- Next ----------------
def next_question(btn):
    global index, score

    if options.value == quiz_data[index]["answer"]:
        score += 1

    index += 1
    load_question()

# ---------------- Result ----------------
def show_result():
    clear_output()

    total_time = int(time.time() - start_time)

    result = widgets.HTML(f"""
    <h1 style='color:#22c55e;text-align:center;'>🏆 Quiz Finished</h1>
    <h2 style='text-align:center;'>Score: {score} / {len(quiz_data)}</h2>
    <p style='text-align:center;color:gray;'>Time Taken: {total_time} sec</p>
    """)

    display(result)

# ---------------- Event ----------------
next_btn.on_click(next_question)

# ---------------- Main Card ----------------
card = widgets.VBox([
    title,
    question,
    options,
    next_btn
])

card.layout.width = '600px'
card.layout.padding = '28px'
card.layout.border_radius = '28px'
card.layout.background = 'linear-gradient(135deg,#eff6ff,#ffffff,#dbeafe)'
card.layout.box_shadow = '0 18px 45px rgba(0,0,0,0.18)'

# ---------------- Show ----------------
display(card)
load_question()
```
