```python
# Attractive Calculator Web Application in Python (Google Colab / Jupyter)

import ipywidgets as widgets
from IPython.display import display

# ---------------- Display Screen ----------------
screen = widgets.Text(
    value='',
    placeholder='0',
    disabled=False
)

screen.layout.width = '360px'
screen.layout.height = '55px'
screen.style.description_width = '0px'

# ---------------- Functions ----------------
def press(value):
    screen.value += str(value)

def clear(btn):
    screen.value = ''

def equal(btn):
    try:
        screen.value = str(eval(screen.value))
    except:
        screen.value = "Error"

# ---------------- Button Maker ----------------
def make_button(text, color=""):
    btn = widgets.Button(description=text)
    btn.layout.width = '80px'
    btn.layout.height = '55px'
    btn.layout.margin = '4px'
    btn.style.button_color = color
    return btn

# ---------------- Buttons ----------------
layout_values = [
    ['7','8','9','/'],
    ['4','5','6','*'],
    ['1','2','3','-'],
    ['0','.','=','+'],
    ['C']
]

rows = []

for row in layout_values:
    row_buttons = []

    for item in row:
        if item == '=':
            btn = make_button(item, '#22c55e')
            btn.on_click(equal)
        elif item == 'C':
            btn = make_button(item, '#ef4444')
            btn.on_click(clear)
        elif item in ['+','-','*','/']:
            btn = make_button(item, '#3b82f6')
            btn.on_click(lambda x, val=item: press(val))
        else:
            btn = make_button(item, '#e2e8f0')
            btn.on_click(lambda x, val=item: press(val))

        row_buttons.append(btn)

    rows.append(widgets.HBox(row_buttons))

# ---------------- Title ----------------
title = widgets.HTML("""
<h1 style="
text-align:center;
font-size:34px;
background:linear-gradient(90deg,#06b6d4,#2563eb,#8b5cf6);
-webkit-background-clip:text;
-webkit-text-fill-color:transparent;
margin-bottom:10px;">
🧮 Luxury Calculator
</h1>
""")

# ---------------- Footer ----------------
footer = widgets.HTML("""
<center>
<p style='color:gray;font-style:italic;'>
Simple • Elegant • Fast
</p>
</center>
""")

# ---------------- Main Card ----------------
app = widgets.VBox([title, screen] + rows + [footer])

app.layout.width = '390px'
app.layout.padding = '25px'
app.layout.border_radius = '25px'
app.layout.box_shadow = '0 15px 40px rgba(0,0,0,0.18)'
app.layout.background = 'linear-gradient(135deg,#eff6ff,#ffffff,#dbeafe)'

# ---------------- Display ----------------
display(app)
```
