```python
import ipywidgets as widgets
from IPython.display import display
```

```python
logo = widgets.HTML(value="<h2 style='margin:0;color:#2563eb;'>🚀 SkillCraft</h2>")
menu = widgets.HBox([
    widgets.Button(description='Home'),
    widgets.Button(description='About'),
    widgets.Button(description='Services'),
    widgets.Button(description='Contact')
])
for b in menu.children:
    b.layout.width='110px'
    b.style.button_color='#3b82f6'
    b.style.text_color='white'

header = widgets.HBox([logo, menu])
header.layout.justify_content='space-between'
header.layout.align_items='center'
header.layout.padding='15px 25px'
header.layout.border='1px solid #dbeafe'

# Hero Section
hero_title = widgets.HTML(value="""
<h1 style='font-size:42px;color:#111827;margin:10px 0;'>Launch Extraordinary Digital Experiences</h1>
<p style='font-size:18px;color:#475569;'>Modern responsive landing page built entirely with Python widgets.</p>
""")
cta = widgets.Button(description='Get Started')
cta.style.button_color='#06b6d4'
cta.style.text_color='white'
cta.layout.width='180px'
cta.layout.height='45px'
hero = widgets.VBox([hero_title, cta])
hero.layout.align_items='center'
hero.layout.padding='50px'
hero.layout.margin='20px 0'
hero.layout.border='1px solid #e2e8f0'

# Cards
cards=[]
items=[('🌐 Web Development','Scalable modern websites'),('🎨 UI / UX Design','Beautiful user experiences'),('📱 Responsive Design','Works on every device'),('🚀 SEO Growth','Speed and visibility')]
for title,desc in items:
    box=widgets.VBox([
        widgets.HTML(value=f"<h3>{title}</h3>"),
        widgets.HTML(value=f"<p style='color:#475569'>{desc}</p>")
    ])
    box.layout.padding='20px'
    box.layout.border='1px solid #cbd5e1'
    box.layout.border_radius='12px'
    box.layout.width='240px'
    cards.append(box)
services = widgets.HBox(cards)
services.layout.justify_content='space-around'
services.layout.flex_flow='row wrap'
services.layout.gap='15px'

# Footer
footer = widgets.HTML(value="<center><b>© 2026 SkillCraft Technology • Built with Python</b></center>")

# Main Layout
page = widgets.VBox([header, hero, widgets.HTML('<h2>Our Services</h2>'), services, footer])
page.layout.padding='20px'
page.layout.width='100%'
page.layout.border='2px solid #bfdbfe'

# Display page
display(page)
print('Pure Python landing page rendered successfully.')
```
